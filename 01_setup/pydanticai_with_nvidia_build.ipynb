# PydanticAI with Nvidia Build

```python
from pydantic_ai import Agent
from pydantic_ai.models.openai import OpenAIChatModel
from pydantic_ai.providers.openai import OpenAIProvider
import os
from dotenv import load_dotenv

load_dotenv()

model = OpenAIChatModel(
    model_name = "nvidia/nemotron-3.5-lightning-30b-a3b",
    provider = OpenAIProvider(  
        base_url = "https://integrate.api.nvidia.com/v1", api_key =os.environ['Nvidia_API']
        ),
)

model
```

```python
agent = Agent(model)
result = await agent.run("Give me a dad joke in data engineering", timeout=60)
result
```

## Change Model

##### ändrar model genom att kopiera "model=.." från Nvidia Models. Man kan kolla igenom vilka språk modellen kör i Model card

```python
model= OpenAIChatModel(
    model_name="nvidia/nemotron-3.5-super-vl-120b-a12b",
    provider=OpenAIProvider(
        base_url="https://integrate.api.nvidia.com/v1", api_key=os.environ['Nvidia_API']
    ),
)

result = await agent.run("säg en pappa skämt", model=model, timeout=60)
result
```
